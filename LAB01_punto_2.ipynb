EJERCICIO 1

```python
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ").strip().lower()
no_sancionado = input("¿Está libre de sanciones viales? (s/n): ").strip().lower()

if edad >= 18 and tiene_licencia == 's' and no_sancionado == 's':
    print("¡Felicidades! Puedes conducir legalmente.")
else:
    print("Lo sentimos, no cumples con los requisitos para conducir (Mayor de 18 años, licencia vigente y sin sanciones).")
```

EJERCICIO 2

```python
n = int(input("Ingrese un número n (límite superior): "))
suma_total = 0

print(f"Números pares entre 1 y {n}:")
for i in range(1, n + 1):
    if i % 2 == 0:
        print(f"  - {i} es par")

    suma_total += i

print(f"La suma total de los primeros {n} números naturales es: {suma_total}")
```

EJERCICIO 3

```python
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1, 26.4]

promedio = sum(temperaturas) / len(temperaturas)
print(f"Temperatura promedio: {promedio:.2f}°C")

mayores_al_promedio = [t for t in temperaturas if t > promedio]
print("Temperaturas mayores al promedio:", mayores_al_promedio)
```

EJERCICIO 4

```python
contador = 5
print("Cuenta atrás:")
while contador > 0:
    print("Contador:", contador)
    contador -= 1
print("¡Despegue!\n")

numero = int(input("Ingrese un número (ingrese 0 para terminar): "))
while numero != 0:
    print(f"Número registrado: {numero}")
    numero = int(input("Ingrese otro número (0 para salir): "))

print("Programa finalizado.")
```

EJERCICIO 5

```python
suma_pares = 0
suma_impares = 0

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

estado_final = {
    "suma_pares": suma_pares,
    "suma_impares": suma_impares
}

print("Estado final del proceso:", estado_final)
```

EJERCICIO 6

```python
productos = []

while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre.lower() == "fin":
        break

    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))

    if precio < 0 or cantidad < 0:
        print("❌ Error: El precio o la cantidad no pueden ser negativos. Producto omitido.")
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = sum(p["precio"] * p["cantidad"] for p in productos)
print(f"\nTotal del inventario: ${total:,.2f}")

if productos:
    producto_mas_caro = max(productos, key=lambda x: x["precio"])
    print(f"👑 Producto más caro: {producto_mas_caro['nombre']} (${producto_mas_caro['precio']})")
else:
    print("No se ingresaron productos válidos.")

print("\nEstado final del inventario (Lista de diccionarios):")
print(productos)
```
