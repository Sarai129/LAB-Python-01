```python
def agregar_nota(notas, nueva):

    print("ID antes append:", id(notas))

    notas.append(nueva)

    print("ID después append:", id(notas))


def reiniciar(notas):

    print("ID antes reasignar:", id(notas))

    notas = []

    print("ID después reasignar:", id(notas))


mis_notas = [3.5, 4.0]

print("ID original:", id(mis_notas))

agregar_nota(mis_notas, 4.8)

reiniciar(mis_notas)
```

```python
def consultar(cuenta):

    print("\nTitular:", cuenta["titular"])
    print("Saldo:", cuenta["saldo"])


def depositar(cuenta, monto):

    cuenta["saldo"] += monto

    movimiento = {
        "tipo": "Deposito",
        "monto": monto,
        "saldo": cuenta["saldo"]
    }

    cuenta["movimientos"].append(movimiento)

    print("Depósito realizado")


def retirar(cuenta, monto):

    if monto > cuenta["saldo"]:
        print("Saldo insuficiente")
        return

    cuenta["saldo"] -= monto

    movimiento = {
        "tipo": "Retiro",
        "monto": monto,
        "saldo": cuenta["saldo"]
    }

    cuenta["movimientos"].append(movimiento)

    print("Retiro realizado")
```

```python
cuenta = {
    "titular": "Ana",
    "saldo": 200000,
    "movimientos": []
}

while True:

    print("\n===== CAJERO =====")
    print("1. Consultar")
    print("2. Depositar")
    print("3. Retirar")
    print("4. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":

        consultar(cuenta)

    elif opcion == "2":

        monto = float(input("Monto a depositar: "))
        depositar(cuenta, monto)

    elif opcion == "3":

        monto = float(input("Monto a retirar: "))
        retirar(cuenta, monto)

    elif opcion == "4":
        break

    else:
        print("Opción inválida")
```

```python
print("\n===== EXTRACTO =====")

print("Tipo\t\tMonto\t\tSaldo")

for movimiento in cuenta["movimientos"]:

    print(
        movimiento["tipo"],
        "\t",
        movimiento["monto"],
        "\t",
        movimiento["saldo"]
    )
```
