```python
# Entrada de valores por teclado y conversión a float
saldo = float(input("Ingrese el saldo inicial: "))
deposito = float(input("Ingrese el monto a depositar: "))
retiro = float(input("Ingrese el monto a retirar: "))

# Declaración de la variable de interés
interes = 0.02  # 2% mensual

saldo = saldo + deposito
saldo = saldo - retiro

saldo_final = saldo + (saldo * interes)

# Salida del estado final usando un f-string
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

EJERCICIO 3

```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

print("-------------")
print(inventario)
print("-" * 30)
print(f"Nombre del primer producto: {inventario[0]['nombre']}")


nuevo_producto = {"nombre": "Huevos", "precio": 6000, "stock": 8}
inventario.append(nuevo_producto)
print(f"Producto agregado: {nuevo_producto['nombre']}")

for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] -= 2
print("Stock de Leche actualizado tras venta (-2 unidades).")

valor_total = sum(p["precio"] * p["stock"] for p in inventario)
print(f"Valor total del inventario: ${valor_total:,.2f}")

print("Productos con stock menor a 10:")
for p in inventario:
    if p["stock"] < 10:
        print(f"   - {p['nombre']}: {p['stock']} unidades")

for p in inventario:
    p["precio"] = round(p["precio"] * 1.10, 2)
print(" Precios aumentados en un 10% y redondeados.")


busqueda = input("\n Ingrese el nombre del producto que desea buscar: ").strip().capitalize()
encontrado = None
for p in inventario:
    if p["nombre"].lower() == busqueda.lower():
        encontrado = p
        break

if encontrado:
    print(f"   ¡Encontrado! -> Producto: {encontrado['nombre']} | Precio: ${encontrado['precio']} | Stock: {encontrado['stock']}")
else:
    print(f"   Lo sentimos, el producto '{busqueda}' no se encuentra registrado en el inventario.")

inventario[0]["stock"] = 0

inventario = [p for p in inventario if p["stock"] > 0]
print("\n Productos con stock 0 eliminados del inventario.")

inventario.sort(key=lambda x: x["precio"])
print("\n Inventario final ordenado por precio ascendente:")
for p in inventario:
    print(f"    - {p['nombre']}: ${p['precio']} (Stock: {p['stock']})")
```

EJERCICIO 4

```python
import math

punto = (3, 7)
print("Punto:", punto)
print("-" * 40)

rectangulo = (5, 10)
print(" Tupla rectángulo (ancho, alto):", rectangulo)

ancho, alto = rectangulo
area = ancho * alto
perimetro = 2 * (ancho + alto)
print(f"Área: {area} | Perímetro: {perimetro}")

posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0.0

for i in range(len(posiciones) - 1):
    x1, y1 = posiciones[i]
    x2, y2 = posiciones[i + 1]
    distancia_tramo = math.hypot(x2 - x1, y2 - y1)
    distancia_total += distancia_tramo

print(f"\nDistancia total recorrida entre {posiciones}: {distancia_total:.2f}")
```

EJERCICIO 5

```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

while True:
    print("\n--- SISTEMA DE BIBLIOTECA ---")
    print("1. Ver estado y conteo de libros (Prestados / Disponibles)")
    print("2. Agregar un nuevo libro")
    print("3. Prestar o devolver un libro")
    print("4. Ver estado completo del sistema (print)")
    print("5. Salir")

    opcion = input("Selecciona una opción (1-5): ").strip()

    if opcion == "1":
        prestados = sum(1 for libro in biblioteca["libros"] if libro["prestado"])
        disponibles = len(biblioteca["libros"]) - prestados
        print(f"\n📊 [CONTEO]")
        print(f"   - Libros prestados: {prestados}")
        print(f"   - Libros disponibles: {disponibles}")
        print(f"   - Total de libros: {len(biblioteca['libros'])}")

    elif opcion == "2":
        print("\n➕ [AGREGAR LIBRO]")
        titulo = input("Ingresa el título del libro: ").strip()
        autor = input("Ingresa el autor del libro: ").strip()

        if titulo and autor:
            nuevo_libro = {"titulo": titulo, "autor": autor, "prestado": False}
            biblioteca["libros"].append(nuevo_libro)
            print(f"¡Libro '{titulo}' agregado exitosamente!")
        else:
            print("❌ Error: El título y el autor no pueden estar vacíos.")

    elif opcion == "3":
        print("\n📖 [GESTIÓN DE PRÉSTAMOS]")
        busqueda = input("Ingresa el título del libro que deseas gestionar: ").strip().lower()
        encontrado = False

        for libro in biblioteca["libros"]:
            if libro["titulo"].lower() == busqueda:
                encontrado = True
                print(f"Libro encontrado: '{libro['titulo']}' (Estado actual prestado: {libro['prestado']})")
                accion = input("¿Qué deseas hacer? (1: Marcar como prestado, 2: Marcar como devuelto/disponible): ").strip()

                if accion == "1":
                    libro["prestado"] = True
                    print(f"✅ El libro '{libro['titulo']}' ahora figura como PRESTADO.")
                elif accion == "2":
                    libro["prestado"] = False
                    print(f"✅ El libro '{libro['titulo']}' ahora figura como DISPONIBLE.")
                else:
                    print("⚠️ Opción no válida.")
                break

        if not encontrado:
            print(f"❌ Lo sentimos, el libro no se encuentra en el inventario.")

    elif opcion == "4":
        print("\n📦 [ESTADO COMPLETO DEL SISTEMA]")
        print(biblioteca)

    elif opcion == "5":
        print("\n👋 ¡Gracias por usar el sistema de la biblioteca. Hasta luego!")
        break
    else:
        print("❌ Opción inválida. Por favor, selecciona un número entre 1 y 5.")
```
