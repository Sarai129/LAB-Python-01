EJERCICIO 1

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()

nombres_biz = []
for u in usuarios:
    if u.get("email", "").endswith(".biz"):
        nombres_biz.append(u["name"])

print("Usuarios encontrados con correo terminado en .biz:")
if nombres_biz:
    for n in nombres_biz:
        print("-", n)
else:
    print("No se encontraron usuarios con ese dominio.")
```

EJERCICIO 2

```python
import requests

nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").strip().lower()

url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    datos = respuesta.json()

    tipos = []
    for t in datos["types"]:
        tipos.append(t["type"]["name"])

    habilidades = []
    for h in datos["abilities"]:
        habilidades.append(h["ability"]["name"])

    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": tipos,
        "habilidades": habilidades,
        "estadisticas": datos["stats"]
    }

    print("\n📘 Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")
    print(f"Altura: {pokemon['altura']}")
    print(f"Peso: {pokemon['peso']}")
    print(f"Tipos: {', '.join(pokemon['tipos'])}")

    if len(pokemon["tipos"]) > 1:
        print("🌟 Característica: ¡Pokémon de tipo mixto!")

    if pokemon["peso"] > 100:
        print("⚖️ Clasificación: Pokémon de gran tamaño")
    else:
        print("⚖️ Clasificación: Pokémon ligero")

    print("\nHabilidades registradas:")
    for hab in pokemon["habilidades"]:
        print(f"  • {hab.capitalize()}")

    print("\n📊 Estadísticas Base:")
    for stat in pokemon["estadisticas"]:
        nombre_stat = stat["stat"]["name"].upper()
        valor_stat = stat["base_stat"]
        print(f"  - {nombre_stat}: {valor_stat}")

else:
    print("❌ Pokémon no encontrado. Verifique el nombre ingresado.")
```
