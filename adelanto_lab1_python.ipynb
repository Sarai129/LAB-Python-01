```python
# Estado inicial
saldo = float(input("Digite el saldo de su cuenta: ")) # Saldo de la cuenta
estado = input("¿Qué opcion desea realizar?. Depositar o retirar (D/R): ")
if estado == "D" or "d":
  deposito = float(input("Digite el monto a depositar: "))   # Monto a depositar
elif estado == "R" or "r":
  retiro = float(input("Digite el monto a retirar: "))       # Monto a retirar

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero
interes = 0.02 #(2% mensual)
interes_saldo = saldo*interes
saldo_final = saldo + interes_saldo

# Salida del estado final
print("Saldo final de la cuenta:", saldo)
print(f"Saldo final con interés: ${saldo_final:.2f}")
```

```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break
    precio = float(input("Precio: "))
    if precio <=0:
      continue
    cantidad = int(input("Cantidad: "))
    if cantidad <= 0:
      continue
    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

# Mostrar total
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)
```
