```python
import random

ganadas = 0
perdidas = 0

seguir = "s"

while seguir.lower() == "s":

    numero_secreto = random.randint(1, 100)
    historial = []
    adivino = False

    print("\n=== JUEGO DE ADIVINANZA ===")
    print("Debes adivinar un número entre 1 y 100")

    intentos = 1

    while intentos <= 7:

        numero = int(input(f"Intento {intentos}: "))

        historial.append(numero)

        if numero == numero_secreto:
            print("¡Correcto!")
            ganadas += 1
            adivino = True
            break

        elif numero < numero_secreto:
            print("Mayor")

        else:
            print("Menor")

        intentos += 1

    if not adivino:
        print("Perdiste.")
        print("El número era:", numero_secreto)
        perdidas += 1

    print("Historial de intentos:", historial)

    seguir = input("¿Desea jugar otra vez? (s/n): ")

print("\n=== MARCADOR FINAL ===")
print("Partidas ganadas:", ganadas)
print("Partidas perdidas:", perdidas)
```

```python
import random

tablero = []

for fila in range(10):

    nueva_fila = []

    for columna in range(10):
        nueva_fila.append(random.randint(0, 1))

    tablero.append(nueva_fila)
```

```python
def mostrar_tablero(tablero):

    for fila in tablero:

        for celda in fila:

            if celda == 1:
                print("■", end=" ")
            else:
                print("□", end=" ")

        print()

    print()
```

```python
def contar_vecinos(tablero, fila, columna):

    vecinos = 0

    for i in range(fila - 1, fila + 2):

        for j in range(columna - 1, columna + 2):

            if i == fila and j == columna:
                continue

            if i >= 0 and i < 10 and j >= 0 and j < 10:

                if tablero[i][j] == 1:
                    vecinos += 1

    return vecinos
```

```python
def siguiente_generacion(tablero):

    nuevo_tablero = []

    for fila in range(10):

        nueva_fila = []

        for columna in range(10):

            vecinos = contar_vecinos(
                tablero,
                fila,
                columna
            )

            if tablero[fila][columna] == 1:

                if vecinos == 2 or vecinos == 3:
                    nueva_fila.append(1)
                else:
                    nueva_fila.append(0)

            else:

                if vecinos == 3:
                    nueva_fila.append(1)
                else:
                    nueva_fila.append(0)

        nuevo_tablero.append(nueva_fila)

    return nuevo_tablero
```

```python
print("GENERACIÓN 0")
mostrar_tablero(tablero)

for generacion in range(1, 6):

    tablero = siguiente_generacion(tablero)

    print("GENERACIÓN", generacion)

    mostrar_tablero(tablero)
```
